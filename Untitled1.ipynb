```python
import os
import cv2
import numpy as np

print("1. Chargement des données avec Sobel Gradients...")

my_path = r"C:\Users\hp\Desktop\projettrimage\dataset"
cats_dir = os.path.join(my_path, 'cats')
dogs_dir = os.path.join(my_path, 'dogs')

X = []
Y = []

def load_images_sobel(folder, label):
    if os.path.exists(folder):
        for img_name in os.listdir(folder):
            img_path = os.path.join(folder, img_name)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            if img is not None:
                # 1. Redimensionner l'image en 64x64
                img_resized = cv2.resize(img, (64, 64))
                
                # 2. Floutage pour réduire le bruit
                blurred = cv2.GaussianBlur(img_resized, (3, 3), 0)
                
                # 3. Calcul des gradients Sobel X et Y
                sobelx = cv2.Sobel(blurred, cv2.CV_64F, 1, 0, ksize=3)
                sobely = cv2.Sobel(blurred, cv2.CV_64F, 0, 1, ksize=3)
                
                # 4. Magnitude du gradient
                magnitude = np.sqrt(sobelx**2 + sobely**2)
                
                # 5. Normalisation
                magnitude = cv2.normalize(magnitude, None, 0, 1, cv2.NORM_MINMAX)
                
                X.append(magnitude.flatten())
                Y.append(label)

load_images_sobel(cats_dir, 0)
load_images_sobel(dogs_dir, 1)

X = np.array(X)
Y = np.array(Y).reshape(-1, 1)

print(f"Extraction terminée! Images traitées : {X.shape[0]}")

# Split Train / Test (80% / 20%)
np.random.seed(42)
indices = np.arange(X.shape[0])
np.random.shuffle(indices)

split_idx = int(0.8 * X.shape[0])
train_idx, test_idx = indices[:split_idx], indices[split_idx:]

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = Y[train_idx], Y[test_idx]

# =====================================================================
# MLP optimisé avec Sobel
# =====================================================================
print("\nEntraînement en cours...")

input_size = X_train.shape[1]
hidden_size = 64  # Taille moyenne pour éviter l'overfitting
output_size = 1

# Initialisation de He
W1 = np.random.randn(input_size, hidden_size) * np.sqrt(2.0 / input_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size) * np.sqrt(2.0 / hidden_size)
b2 = np.zeros((1, output_size))

lr = 0.0005  # Learning rate plus petit pour un apprentissage stable
epochs = 800
lambda_reg = 0.02

def sigmoid(z):
    return 1 / (1 + np.exp(-np.clip(z, -500, 500)))

def sigmoid_derivative(z):
    return z * (1 - z)

# Boucle d'entraînement
for epoch in range(epochs):
    # Forward
    Z1 = np.dot(X_train, W1) + b1
    A1 = sigmoid(Z1)
    Z2 = np.dot(A1, W2) + b2
    A2 = sigmoid(Z2)
    
    # Backprop
    error = y_train - A2
    dZ2 = error * sigmoid_derivative(A2)
    
    dW2 = np.dot(A1.T, dZ2) - (lambda_reg * W2)
    db2 = np.sum(dZ2, axis=0, keepdims=True)
    
    dZ1 = np.dot(dZ2, W2.T) * sigmoid_derivative(A1)
    dW1 = np.dot(X_train.T, dZ1) - (lambda_reg * W1)
    db1 = np.sum(dZ1, axis=0, keepdims=True)
    
    W2 += dW2 * lr
    b2 += db2 * lr
    W1 += dW1 * lr
    b1 += db1 * lr

# Test
Z1_test = np.dot(X_test, W1) + b1
A1_test = sigmoid(Z1_test)
Z2_test = np.dot(A1_test, W2) + b2
predictions = sigmoid(Z2_test)

y_pred = (predictions > 0.5).astype(int)
accuracy = np.mean(y_pred == y_test)

print("\n================ Résultat avec Sobel ================")
print(f"Précision du modèle sur le Test : {accuracy * 100:.2f}%")
```
